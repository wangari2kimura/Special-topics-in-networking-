```python
!pip install netmiko

from netmiko import ConnectHandler
import time

device = {
    'device_type': 'cisco_ios',
    'host': 'devnetsandboxiosxec8k.cisco.com', # <-- RE-VERIFY THIS IN DASHBOARD
    'username': 'vivian.kimura',
    'password': 'W_Q6nc1x_3_6G',
    'port': 22,
    'global_delay_factor': 2, # Tells Netmiko to wait 2x longer than usual
    'fast_cli': False,        # More stable for slow virtual routers
}

try:
    print(f"Connecting to {device['host']}... (this may take a moment)")
    # We add a longer timeout for the initial connection
    net_connect = ConnectHandler(**device, conn_timeout=60)

    print("Successfully logged in!")
    print("\n--- Interface Status ---")
    output = net_connect.send_command('show ip int brief')
    print(output)

    # Configuration part
    config_commands = [
        'interface Loopback99',
        'description Configured via Python Netmiko',
        'ip address 192.168.99.1 255.255.255.255'
    ]

    print("\nSending configuration commands...")
    # 'read_timeout' gives the router extra time to process the config
    config_output = net_connect.send_config_set(config_commands, read_timeout=30)
    print(config_output)

    net_connect.disconnect()
    print("\nLab Complete: Disconnected successfully.")

except Exception as e:
    print(f"\n[ERROR]: {e}")


```
